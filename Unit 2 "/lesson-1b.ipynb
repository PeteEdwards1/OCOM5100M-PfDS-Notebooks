## Exercise 5

Use `if`, `elif` and `else` to encode the rules "if it is 9am or later I will get up, else if it is 7am or later I will read my book in bed, otherwise I will go back to sleep", which determine what to do depending on the time. The starter code below creates a random time as a string, and sets the value of the variable message depending on the `if` statement. Add `elif` and else statements to set message to "stay in bed and read your book." if the time is 7 or later and "go back to sleep." otherwise.

### Solution 5

We use `int(time[:2])` to extract the hour from the time string and cast it to an integer. We then use >= since any time starting with 09 is 9am or later. We can do the same for the 7am condition.

```python
import random

# randint generates a random integer within the given range.
# The f-string format :0>2 ensures the hour and minute are always two digits
# (e.g. 07, 09) so that the time string always looks like HH:MM.
# Note: hours are limited to 0-11 to keep the output in the morning.
time = f'{random.randint(0,11):0>2}:{random.randint(0,59):0>2}'

# message is initialised here so it always has a value,
# even before the conditions below are evaluated.
message = "go back to sleep."

if int(time[:2]) >= 9:
    message = "time to get up!"
elif int(time[:2]) >= 7:
    message = "stay in bed and read your book."
else:
    message = "go back to sleep."

print(f'The time is {time}, {message}')
```

### Solution 6

We cannot use `int(time[:2]) > 9` because this only compares the hour, and would incorrectly exclude times like `09:01`. Instead, we compare the full time string directly using `time > '09:00'`, which correctly identifies any time strictly after `09:00`.


```python
import random

# It's worth setting the time ourselves here as the edge case is unlikely
# to occur with a random time.
# time = f'{random.randint(0,11):0>2}:{random.randint(0,59):0>2}'
time = '09:00'

if time > '09:00':
    message = "time to get up!"
elif time > '07:00':
    message = "stay in bed and read your book."
else:
    message = "go back to sleep."

print(f'The time is {time}, {message}')
```

## Exercise 7

A supermarket is offering a discount to people who are either over 65 or members of the supermarket's rewards scheme. First, use an `if` statement to evaluate whether the customer's age is over 65, otherwise determine if they are a member. If the customer is over 65 or is a member, set the `discount_eligible` variable to `True`.

### Solution 7

There are multiple ways to complete the objective of this task. One way using a nested `if` statement is:



```python
import random

# This code randomises the customer's age and membership status.
# Each customer has a 50% chance of being a member, regardless of age.
age = random.randint(10, 99)
is_member = random.random() < 0.5

discount_eligible = False
if age > 65:
    discount_eligible = True
else:
    if is_member:
        discount_eligible = True

print(f'Age: {age}, Member: {is_member}, Eligible for discount: {discount_eligible}')
```

## Exercise 8

Was it necessary to include a nested `if` statement to achieve the goal set out in the previous exercise? If not, can you write a program that does it without using a nested `if` statement? You might consider the advantages and disadvantages of both programs.

### Solution 8

It is not necessary to use a nested `if` statement. The same result can be achieved more concisely using the `or` operator:


```python
import random

# This code randomises the customer's age and membership status.
# Each customer has a 50% chance of being a member, regardless of age.
age = random.randint(10, 99)
is_member = random.random() < 0.5

discount_eligible = False
if age > 65 or is_member:
    discount_eligible = True

print(f'Age: {age}, Member: {is_member}, Eligible for discount: {discount_eligible}')
```

Both programs produce the same result. The nested version makes the logic more explicit, you can clearly see the two conditions being checked separately. The version using `or` is more concise and easier to read, but requires understanding how Boolean operators work. In general, avoiding unnecessary nesting makes code easier to follow and less prone to bugs.
